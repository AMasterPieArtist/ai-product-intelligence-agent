```python
#尝试调用dify
def call_dify(question: str) -> dict:
    """
    向 Dify 知识库问答应用发送问题，
    返回实际回答、来源信息和执行耗时。
    """

    url = f"{DIFY_BASE_URL}/chat-messages"

    headers = {
        "Authorization": f"Bearer {DIFY_API_KEY}",
        "Content-Type": "application/json"
    }

    payload = {
        "inputs": {},
        "query": question,
        "response_mode": "blocking",
        "conversation_id": "",
        "user": "product-test-user"
    }

    start_time = time.perf_counter()

    response = requests.post(
        url,
        headers=headers,
        json=payload,
        timeout=120
    )

    elapsed_time = time.perf_counter() - start_time

    # 如果 HTTP 请求失败，抛出异常
    response.raise_for_status()

    # 将 API 返回的 JSON 转成 Python 字典
    data = response.json()

    # 获取模型回答
    answer = data.get("answer", "")

    # 获取返回的元数据
    metadata = data.get("metadata") or {}

    # 尝试提取知识库引用
    references = metadata.get("retriever_resources") or []

    return {
        "question": question,
        "answer": answer,
        "references": references,
        "elapsed_time": round(elapsed_time, 2),
        "conversation_id": data.get("conversation_id"),
        "message_id": data.get("message_id"),
        "raw_response": data
    }
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class ProductTestState(TypedDict):
    test_cases: list[dict]
    execution_results: list[dict]
    evaluation_results: list[dict]
    final_report: str
```

```python
#定义节点
def testing_executor_node(state: ProductTestState) -> dict:
    """逐道调用 Dify，并保留真实执行结果。"""

    results = []

    for case in state["test_cases"]:
        try:
            actual = call_dify(case["question"])

            results.append({
                "test_id": case["test_id"],
                "expected_answer": case["expected_answer"],
                "answer_exists": case["answer_exists"],
                "execution_status": "success",
                **actual,
            })

        except Exception as e:
            results.append({
                "test_id": case["test_id"],
                "question": case["question"],
                "expected_answer": case["expected_answer"],
                "answer_exists": case["answer_exists"],
                "execution_status": "failed",
                "answer": "",
                "references": [],
                "elapsed_time": None,
                "error_message": str(e),
            })

    return {"execution_results": results}


def evaluator_node(state: ProductTestState) -> dict:
    """评测已执行的测试，不重新调用 Dify。"""

    results = []

    for result in state["execution_results"]:
        if result["execution_status"] == "failed":
            results.append({
                "test_id": result["test_id"],
                "answer_correct": False,
                "citation_correct": None,
                "evaluation_reason": result["error_message"],
            })
            continue

        evaluation = evaluate_test_result(result)

        results.append({
            "test_id": result["test_id"],
            **evaluation,
        })

    return {"evaluation_results": results}


def report_generator_node(state: ProductTestState) -> dict:
    """根据执行记录和评测结果生成简要报告。"""

    lines = ["# Dify 知识库问答测试报告", ""]

    for result in state["evaluation_results"]:
        lines.extend([
            f"## {result['test_id']}",
            f"- 实际回答：{result.get('actual_answer', '')}",
            f"- 答案检查：{result['answer_correct']}",
            f"- 引用检查：{result['citation_correct']}",
            f"- 执行耗时：{result.get('elapsed_time')} 秒",
            "",
        ])

    return {"final_report": "\n".join(lines)}
```

```python
builder = StateGraph(ProductTestState)

builder.add_node("testing_executor", testing_executor_node)
builder.add_node("evaluator", evaluator_node)
builder.add_node("report_generator", report_generator_node)

builder.add_edge(START, "testing_executor")
builder.add_edge("testing_executor", "evaluator")
builder.add_edge("evaluator", "report_generator")
builder.add_edge("report_generator", END)

product_test_graph = builder.compile()
```

```python
#运行完整工作流
initial_state = {
    "test_cases": test_questions,
    "execution_results": [],
    "evaluation_results": [],
    "final_report": "",
}

final_state = product_test_graph.invoke(initial_state)

print(final_state["final_report"])
```

```python
from typing import Literal
from pydantic import BaseModel, Field


class ExecutableTestCase(BaseModel):
    test_id: str
    scenario: Literal["fact_qa", "missing_information"]
    question: str

    # fact_qa：应当出现在回答中的事实
    # missing_information：应为空列表
    expected_facts: list[str] = Field(default_factory=list)

    required_source: str


class ExecutableTestPlan(BaseModel):
    test_cases: list[ExecutableTestCase]
```

```python
#让planner生成可执行的测试题
import json

executable_planner = model.with_structured_output(
    ExecutableTestPlan
)

planner_prompt = f"""
你是一名 AI 产品测试设计员。

请根据以下已知测试资料，生成恰好 3 道可通过聊天 API
直接执行的中文测试题：

1. 两道 fact_qa：考查文档中明确存在的事实。
2. 一道 missing_information：考查文档未提供的信息。

严格要求：
- 只能使用提供的资料设计题目和预期事实。
- 不得新增人物、数字、日期、功能或其他背景事实。
- fact_qa 的 expected_facts 只能填写已知事实中的原文值。
- missing_information 的 expected_facts 必须是空列表。
- question 必须是一条可以直接发送给聊天应用的问题，
  不能是“检查引用是否正确”之类的操作指令。
- required_source 一律填写 sample_knowledge.pdf。
- test_id 依次为 TC001、TC002、TC003。
- 不要把参考答案直接写进 question。

测试资料：
{json.dumps(test_spec, ensure_ascii=False, indent=2)}
"""

generated_plan = executable_planner.invoke(planner_prompt)

for case in generated_plan.test_cases:
    print(case.model_dump())
```

```python
#校验planner的输出
def validate_test_plan(plan: ExecutableTestPlan, spec: dict):
    cases = plan.test_cases

    assert len(cases) == 3, "测试用例数量不是 3"

    scenarios = [case.scenario for case in cases]

    assert scenarios.count("fact_qa") == 2
    assert scenarios.count("missing_information") == 1

    assert len({case.test_id for case in cases}) == 3, "测试编号重复"

    allowed_facts = set(spec["known_facts"].values())

    for case in cases:
        assert case.question.strip(), f"{case.test_id} 缺少问题"

        assert (
            case.required_source == spec["source_document"]
        ), f"{case.test_id} 来源文档不正确"

        if case.scenario == "fact_qa":
            assert case.expected_facts, (
                f"{case.test_id} 缺少预期事实"
            )

            assert all(
                fact in allowed_facts
                for fact in case.expected_facts
            ), f"{case.test_id} 包含未经确认的参考答案"

        else:
            assert not case.expected_facts, (
                f"{case.test_id} 无答案场景不应填写预期事实"
            )

    return True


validate_test_plan(generated_plan, test_spec)

print("测试计划通过基础校验，可以进入执行阶段。")
```

```python
#建立新的evaluator
import re


def clean_answer(raw_answer: str) -> str:
    """移除完整的 <think>...</think> 标签内容。"""
    return re.sub(
        r"<think>.*?</think>",
        "",
        raw_answer,
        flags=re.DOTALL
    ).strip()


def evaluate_executable_case(
    case: dict,
    execution: dict
) -> dict:
    """
    根据测试用例的 scenario 选择评测规则。
    不再依赖写死的题目文字。
    """

    if execution["execution_status"] != "success":
        return {
            "test_id": case["test_id"],
            "answer_pass": None,
            "citation_pass": None,
            "evaluation_reason": "API 执行失败，未完成内容评测"
        }

    answer = clean_answer(execution["answer"])
    references = execution.get("references") or []

    if case["scenario"] == "fact_qa":

        # 1. 参考事实是否都出现在最终回答中
        answer_pass = all(
            fact.replace("、", "") in answer.replace("、", "")
            for fact in case["expected_facts"]
        )

        # 2. 来源文档及检索片段是否支持这些事实
        source_text = "\n".join(
            ref.get("content", "")
            for ref in references
            if ref.get("document_name") == case["required_source"]
        )

        citation_pass = bool(source_text) and all(
            fact.replace("、", "") in source_text.replace("、", "")
            for fact in case["expected_facts"]
        )

        reason = (
            "参考事实与检索证据检查完成"
            if answer_pass and citation_pass
            else "回答或检索证据未通过基础事实匹配"
        )

    elif case["scenario"] == "missing_information":

        refusal_keywords = [
            "无法确定",
            "无法得知",
            "未提供",
            "没有提供",
            "无法从",
            "无相关信息",
            "无法回答",
            "没有相关",
            "未包含"
        ]

        has_refusal = any(
            keyword in answer
            for keyword in refusal_keywords
        )

        # 本次只针对营业收入场景检查明显的金额编造
        has_amount = bool(
            re.search(
                r"\d+(?:\.\d+)?\s*(?:亿元|万元|元|美元)",
                answer
            )
        )

        answer_pass = has_refusal and not has_amount

        # 无答案问题不应强行要求引用某段“支持答案”的原文
        citation_pass = None

        reason = (
            "检测到资料不足的说明，未检测到具体金额"
            if answer_pass
            else "未通过基础拒答检查，需要人工复核"
        )

    else:
        raise ValueError(
            f"未知测试场景：{case['scenario']}"
        )

    return {
        "test_id": case["test_id"],
        "answer_pass": answer_pass,
        "citation_pass": citation_pass,
        "evaluation_reason": reason
    }
```

```python
#定义工作流state
from typing import TypedDict
from langgraph.graph import StateGraph, START, END


class DynamicTestState(TypedDict):
    test_cases: list[dict]
    execution_results: list[dict]
    evaluation_results: list[dict]
    final_report: str
```

```python
#定义executor节点
def dynamic_executor_node(state: DynamicTestState) -> dict:

    execution_results = []

    for case in state["test_cases"]:

        print(f"正在执行 {case['test_id']}：{case['question']}")

        try:
            actual = call_dify(case["question"])

            result = {
                "test_id": case["test_id"],
                "execution_status": "success",
                "question": case["question"],
                "answer": actual["answer"],
                "references": actual["references"],
                "elapsed_time": actual["elapsed_time"],
                "message_id": actual["message_id"],
                "error_message": None
            }

        except Exception as e:

            result = {
                "test_id": case["test_id"],
                "execution_status": "failed",
                "question": case["question"],
                "answer": "",
                "references": [],
                "elapsed_time": None,
                "message_id": None,
                "error_message": str(e)
            }

        execution_results.append(result)

    return {
        "execution_results": execution_results
    }
```

```python
#定义evaluator节点
def dynamic_evaluator_node(state: DynamicTestState) -> dict:

    evaluation_results = []

    for case, execution in zip(
        state["test_cases"],
        state["execution_results"]
    ):

        evaluation = evaluate_executable_case(
            case,
            execution
        )

        evaluation_results.append({
            **evaluation,
            "question": case["question"],
            "scenario": case["scenario"],
            "expected_facts": case["expected_facts"],
            "actual_answer": clean_answer(
                execution["answer"]
            ),
            "references": execution["references"],
            "elapsed_time": execution["elapsed_time"],
            "execution_status": execution["execution_status"],
            "error_message": execution["error_message"]
        })

    return {
        "evaluation_results": evaluation_results
    }
```

```python
#定义不同的错误类型，诊断函数
def diagnose_failure(case, execution, evaluation):
    """
    根据执行结果和评测结果，判断失败发生在哪一层。
    """

    # 1. API / 产品调用失败
    if execution["execution_status"] != "success":
        return "execution_failure"

    # 2. fact_qa 场景
    if case["scenario"] == "fact_qa":

        references = execution.get("references") or []

        # 应该有答案，但没有任何检索证据
        if not references:
            return "retrieval_failure"

        # 有检索结果，但答案不正确
        if evaluation["answer_pass"] is False:
            return "answer_failure"

        # 答案正确，但证据不支持
        if evaluation["citation_pass"] is False:
            return "citation_failure"

        return "none"

    # 3. missing_information 场景
    elif case["scenario"] == "missing_information":

        if evaluation["answer_pass"] is False:
            return "answer_failure"

        return "none"

    return "unknown"
```

```python
#创建保存函数
import json
from pathlib import Path
from datetime import datetime


def save_test_run(
    result: dict,
    product_name: str = "Dify",
    product_version: str = "v1"
) -> str:

    # 1. 创建保存目录
    output_dir = Path("../evaluation/runs")
    output_dir.mkdir(parents=True, exist_ok=True)

    # 2. 生成当前时间
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    # 3. 生成 run_id
    run_id = f"{product_name.lower()}_{product_version}_{timestamp}"

    # 4. 汇总一些简单统计
    evaluation_results = result["evaluation_results"]

    total_tests = len(evaluation_results)

    passed_tests = sum(
        1
        for item in evaluation_results
        if item["failure_type"] == "none"
    )

    failed_tests = total_tests - passed_tests

    # 5. 组织最终要保存的数据
    run_record = {
        "run_id": run_id,
        "timestamp": timestamp,
        "product_name": product_name,
        "product_version": product_version,

        "summary": {
            "total_tests": total_tests,
            "passed_tests": passed_tests,
            "failed_tests": failed_tests,
            "pass_rate": (
                round(passed_tests / total_tests, 4)
                if total_tests > 0
                else None
            )
        },

        "test_cases": result["test_cases"],
        "execution_results": result["execution_results"],
        "evaluation_results": result["evaluation_results"]
    }

    # 6. 保存成 JSON 文件
    file_path = output_dir / f"{run_id}.json"

    with open(
        file_path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            run_record,
            f,
            ensure_ascii=False,
            indent=2
        )

    print(f"测试记录已保存：{file_path}")

    return str(file_path)
```

```python
#定义统一接口
class ProductAdapter:
    """
    所有 AI 产品适配器的统一接口。
    """

    def run(self, question: str) -> dict:
        raise NotImplementedError(
            "子类必须实现 run() 方法"
        )
```

```python
#写dify接口
import os
import time
import requests


class DifyAdapter(ProductAdapter):

    def __init__(
        self,
        api_key: str,
        base_url: str
    ):
        self.api_key = api_key
        self.base_url = base_url.rstrip("/")

    def run(self, question: str) -> dict:

        url = f"{self.base_url}/chat-messages"

        headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json"
        }

        payload = {
            "inputs": {},
            "query": question,
            "response_mode": "blocking",
            "conversation_id": "",
            "user": "product-test-user"
        }

        start_time = time.perf_counter()

        try:
            response = requests.post(
                url,
                headers=headers,
                json=payload,
                timeout=120
            )

            elapsed_time = (
                time.perf_counter() - start_time
            )

            response.raise_for_status()

            data = response.json()

            metadata = data.get("metadata") or {}

            references = (
                metadata.get("retriever_resources")
                or []
            )

            return {
                "answer": data.get("answer", ""),
                "references": references,
                "elapsed_time": round(
                    elapsed_time,
                    2
                ),
                "execution_status": "success",
                "message_id": data.get(
                    "message_id"
                ),
                "conversation_id": data.get(
                    "conversation_id"
                ),
                "error_message": None
            }

        except Exception as e:

            elapsed_time = (
                time.perf_counter() - start_time
            )

            return {
                "answer": "",
                "references": [],
                "elapsed_time": round(
                    elapsed_time,
                    2
                ),
                "execution_status": "failed",
                "message_id": None,
                "conversation_id": None,
                "error_message": str(e)
            }
```

稳定性测试 stability test

```python
#编写重复运行函数
from collections import Counter
from statistics import mean


def run_stability_test(
    graph,
    initial_state: dict,
    repeat_times: int = 3,
    product_name: str = "Dify",
    product_version: str = "v1"
):
    """
    连续执行多次完整测试，并保存每一次 Run。
    """

    all_runs = []

    for i in range(repeat_times):

        print(f"\n===== Stability Run {i + 1}/{repeat_times} =====")

        result = graph.invoke(initial_state)

        saved_path = save_test_run(
            result,
            product_name=product_name,
            product_version=product_version
        )

        all_runs.append({
            "run_index": i + 1,
            "result": result,
            "saved_path": saved_path
        })

    return all_runs
```

```python
#正式 完整运行三次
stability_runs = run_stability_test(
    graph=full_test_graph,
    initial_state=initial_state,
    repeat_times=3,
    product_name="Dify",
    product_version="v1"
)
```

```python
#统计test case的稳定性
def analyze_stability(stability_runs):

    case_stats = {}

    for run in stability_runs:

        evaluation_results = (
            run["result"]["evaluation_results"]
        )

        for item in evaluation_results:

            test_id = item["test_id"]

            if test_id not in case_stats:
                case_stats[test_id] = {
                    "scenario": item["scenario"],
                    "question": item["question"],
                    "results": [],
                    "latencies": [],
                    "failure_types": []
                }

            passed = (
                item["failure_type"] == "none"
            )

            case_stats[test_id]["results"].append(
                passed
            )

            if item["elapsed_time"] is not None:
                case_stats[test_id][
                    "latencies"
                ].append(
                    item["elapsed_time"]
                )

            case_stats[test_id][
                "failure_types"
            ].append(
                item["failure_type"]
            )

    return case_stats
```

```python
case_stats = analyze_stability(
    stability_runs
)
```

```python
#打印稳定性报告
for test_id, stats in case_stats.items():

    total_runs = len(
        stats["results"]
    )

    passed_runs = sum(
        stats["results"]
    )

    stability_rate = (
        passed_runs / total_runs
        if total_runs > 0
        else None
    )

    avg_latency = (
        round(mean(stats["latencies"]), 2)
        if stats["latencies"]
        else None
    )

    failure_counter = Counter(
        stats["failure_types"]
    )

    print("=" * 60)

    print("Test ID：", test_id)
    print("Scenario：", stats["scenario"])
    print("Question：", stats["question"])

    print(
        "通过次数：",
        f"{passed_runs}/{total_runs}"
    )

    print(
        "Stability Rate：",
        round(stability_rate, 4)
    )

    print(
        "平均响应时间：",
        avg_latency,
        "秒"
    )

    print(
        "Failure Types：",
        dict(failure_counter)
    )
```

```python
#做总体summary
total_cases = 0
total_passes = 0
all_latencies = []
all_failure_types = []

for stats in case_stats.values():

    total_cases += len(
        stats["results"]
    )

    total_passes += sum(
        stats["results"]
    )

    all_latencies.extend(
        stats["latencies"]
    )

    all_failure_types.extend(
        stats["failure_types"]
    )

overall_pass_rate = (
    total_passes / total_cases
    if total_cases > 0
    else None
)

overall_avg_latency = (
    round(mean(all_latencies), 2)
    if all_latencies
    else None
)

print("\n===== Overall Stability Summary =====")

print("总执行次数：", total_cases)
print("总通过次数：", total_passes)

print(
    "Overall Pass Rate：",
    round(overall_pass_rate, 4)
)

print(
    "平均响应时间：",
    overall_avg_latency,
    "秒"
)

print(
    "Failure Type 分布：",
    dict(
        Counter(
            all_failure_types
        )
    )
)
```
